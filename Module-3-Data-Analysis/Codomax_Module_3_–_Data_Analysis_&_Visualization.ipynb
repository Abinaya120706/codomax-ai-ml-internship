3. Create and Load Dataset

For this project, a sample Student Performance dataset is used.

The dataset contains information about students and their marks in Mathematics, Science, and English. It will be used to demonstrate data exploration, cleaning, analysis, and visualization.

```python
# Create a sample student performance dataset

data = {
    "Student": [
        "Student 1", "Student 2", "Student 3", "Student 4", "Student 5",
        "Student 6", "Student 7", "Student 8", "Student 9", "Student 10",
        "Student 11", "Student 12", "Student 13", "Student 14", "Student 15"
    ],
    "Math": [
        85, 72, 90, 65, 78,
        88, 95, 60, 76, 84,
        70, 92, 81, 68, 89
    ],
    "Science": [
        78, 80, 88, 70, 75,
        91, 94, 65, 79, 82,
        73, 90, 85, 69, 87
    ],
    "English": [
        82, 75, 91, 68, 80,
        85, 96, 62, 77, 86,
        72, 93, 84, 71, 88
    ]
}

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
```

Dataset Dimensions

The "shape" property is used to find the number of rows and columns in the dataset.

```python
# Check number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows removed.")
print("Updated dataset shape:", df.shape)
```

6. Data Analysis

In this section, we calculate average marks, identify the highest-performing students, and compare performance across subjects.

```python
# Check the data types of each column
print(df.dtypes)
```

```python
# Find the student with the highest average score

highest_student = df.loc[df["Average"].idxmax()]

print("Highest-performing student:")
print(highest_student)
```

```python
# Find the student with the lowest average score

lowest_student = df.loc[df["Average"].idxmin()]

print("Lowest-performing student:")
print(lowest_student)
```

```python
# Calculate average score for each subject

subject_averages = df[["Math", "Science", "English"]].mean()

print("Average score for each subject:")
print(subject_averages)
```

7.1 Bar Chart – Average Marks by Subject

A bar chart is used to compare the average performance of students across different subjects.

```python
# Bar chart of average marks by subject

subject_averages.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```

7.2 Line Chart – Student Average Scores

A line chart is used to visualize how the average scores vary across students.

```python
# Line chart of student average scores

plt.figure(figsize=(10, 5))

plt.plot(df["Student"], df["Average"], marker="o")

plt.title("Average Score of Each Student")
plt.xlabel("Student")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

7.3 Pie Chart – Performance Categories

The students are grouped into performance categories based on their average scores.

- Excellent: 90 and above
- Good: 75–89
- Needs Improvement: Below 75

```python
# Create performance categories

def performance_category(score):
    if score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average"].apply(performance_category)

print(df[["Student", "Average", "Performance"]])
```

```python
# Count students in each performance category

performance_counts = df["Performance"].value_counts()

print(performance_counts)
```

```python
# Pie chart of performance categories

plt.figure(figsize=(7, 7))

plt.pie(
    performance_counts.values,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Distribution")
plt.show()
```

7.4 NumPy Analysis

NumPy is used to perform numerical calculations on the students' scores.

```python
# Use NumPy to calculate overall statistics

all_scores = df[["Math", "Science", "English"]].to_numpy()

print("Overall Mean:", round(np.mean(all_scores), 2))
print("Overall Maximum:", np.max(all_scores))
print("Overall Minimum:", np.min(all_scores))
print("Overall Standard Deviation:", round(np.std(all_scores), 2))
```
